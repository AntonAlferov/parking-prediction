```python
from huggingface_hub import hf_hub_download
from roboflow import Roboflow
from ultralytics import YOLO
```

Загрузка весов [YOLO11](https://huggingface.co/Ultralytics/YOLO11) c huggingface, создание модели

```python
hf_hub_download('ultralytics/YOLO11', 'yolo11m.pt')
model_yolo11 = YOLO('yolo11m.pt')
```

Загрузка собственного размеченного датасета по инструкции с Roboflow

```python
rf = Roboflow(api_key="********")
project = rf.workspace("tonik-cukag").project("parking_project")
version = project.version(2)
dataset = version.download("yolov11")
```

Запуск обучения, размер 1280 для читабельности номеров

```python
results = model_yolo11.train(
    data=f'{dataset.location}/data.yaml',
    epochs=50,
    imgsz=1280,
    batch=4,
    patience=10
)
```
